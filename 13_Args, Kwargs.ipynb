```python
def joiner(sep , *args):
    """
    Joins string
    """
    res = ''
    for s in args:
        res = res + s + sep
    return res
```

```python
joiner('@', 'Hello', 'Goodbye', '123','AIRCENTER')
```

```python
joiner('!', 'Hello', 'Goodbye', '123','AIRCENTER','Yechiziezafe')
```

```python
def joiner(sep , *harchiziargument):
    """
    Joins string
    """
    res = ''
    for s in harchiziargument:
        res = res + s + sep
    return res
```

```python
joiner('!', 'Hello', 'Goodbye', '123','AIRCENTER','Yechiziezafe')
```

# Map, Filter

## Map

```python
my_pets = ['alfred', 'tabitha', 'william', 'arla']
my_pets_upper = []

for pet in my_pets:
    new_pet = pet.upper()
    my_pets_upper.append(new_pet)

my_pets_upper
```

```python
my_pets = ['alfred', 'tabitha', 'william', 'arla']

def string_upper(x):
    return x.upper()

my_pets_upper = list(map(string_upper , my_pets))
```

```python
my_pets_upper
```

## lambda

```python
string_upper = lambda a:a.upper()
```

```python
string_upper('Hello')
```

```python
my_pets = ['alfred', 'tabitha', 'william', 'arla']

string_upper = lambda a:a.upper()
my_pets_upper = list(map(string_upper , my_pets))

my_pets_upper
```

```python
my_pets = ['alfred', 'tabitha', 'william', 'arla']

my_pets_upper = list(map(lambda x:x.upper() , my_pets))

my_pets_upper
```
