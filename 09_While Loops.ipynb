```python
def my_function(name, last_name):
    s = "Hello, my name is {}".format(name)
    s2 = "My last name is {}".format(last_name)
    print(s)
    print(s2)
```

```python
my_function('Sajjad', 'FAKHREDDINI')
```

```python
my_function(last_name = 'FAKHREDDINI' , name = 'Sajjad')
```
