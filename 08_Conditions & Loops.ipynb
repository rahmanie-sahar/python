```python
for i in range(10):
    if i == 5:
        break
        
    print(i)

print('Out of the loop')
print('Bluh bluh bluh.')
```

```python
for i in range(10):
    if i == 5:
        continue
        
    print(i)

print('Out of the loop')
print('Bluh bluh bluh.')
```

```python
adjs = ['red', 'big' , 'tasty']
fruits = ['apple' , 'banana' , 'melon']

for sefat in adjs:
    for mive in fruits:
        print(sefat + ' ' + mive)
```
