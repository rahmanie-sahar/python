## Multiple Exceptions

```python
for i in range(5):
    try:
        inp = int(input('Please enter a number : '))
        print(10 / inp)
    except ValueError:
        print('Error! Please enter a integer.')
    except ZeroDivisionError:
        print('Error! Your number should be non-zero')
    
```

## Raise an Exception!

```python
x = -1

if x<0:
    raise Exception("Sorry, no numbers below zero")
    
print(x)
```
