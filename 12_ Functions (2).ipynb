```python
def isPrime(n):
    if n<= 1:
        return False
    if n<= 3:
        return True
    
    if (n%2 == 0):
        return False
    
    i = 3
    
    while i<= n ** 0.5:
        if (n%i == 0):
            return False
        i += 1
    return True
```

# Recursive Functions

## Factorial

```python
# 5! = 5*4*3*2*1
# n! = n * (n-1) * ....* 2 * 1
# n! = n * (n-1)!
# n! = n * (n-1) * (n-2)!
# 0! = 1
```

```python
def factorial(n):
    res = 1
    while n>1:
        res = res * n
        n = n-1
    return res
```

```python
print(factorial(5))
print(factorial(6))
print(6 * factorial(5))
```

```python
def Factorial(n):
    if n == 1 or n == 0:
        return 1
    else:
        return n * Factorial(n-1)
```

```python
Factorial(0)
```

```python
Factorial(1)
```

```python
Factorial(3)
# F(3) >>>> 3 * F(3-1) >>>> 3 * 2 * F(2-1) >>> 3 * 2 * 1
```

```python
Factorial(10)
```

```python
#Fibonacci
#Fn = Fn-1 + Fn-2
#F0 = 0, F1 = 1
#F2 = 1+0 = 1
#F3 = 1+1 = 2
#F4 = 2+1 = 3
#..........
```

```python
def fibbo(n):
    if n == 0: return 0
    if n == 1: return 1
    
    return fibbo(n-1) + fibbo(n-2)
```

```python
for n in range(15):
    print('fibbo({}) = {}'.format(n, fibbo(n)))
```
