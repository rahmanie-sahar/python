### Object Methods

```python
class Person():
    def __init__(self, name, age, gender):
        self.name = name #1st attribute
        self.age = age #2nd attribute
        self.gender = gender #3rd attribute
        print('Hello')
    
    def introduce(self): #A method
        print('My name is {}, and I am {} years old.'.format(self.name, self.age))
```

```python
p2 = Person('Nina', 20, 'Female')
```

```python
p2.name
```

```python
p2.age
```

```python
p2.gender
```

```python
p2.introduce()
```

```python
class Person2():
    def __init__(abc, name, age, gender):
        abc.name = name #1st attribute
        abc.age = age #2nd attribute
        abc.gender = gender #3rd attribute
        print('Hello')
    
    def introduce(abc): #A method
        print('My name is {}, and I am {} years old.'.format(abc.name, abc.age))
```

```python
p3 = Person2('Milad', 21, 'Male')
```

```python
p3.introduce()
```

```python
class Cat():
    def __init__(self, name, age, breed):
        self.name = name
        self.age = age
        self.breed = breed
    
    def meow(self):
        print("Meow! my name is {} and I am {} years old.".format(self.name, self.age))

    def say_breed(self):
        print("My breed is {}".format(self.breed))
```

```python
class Dog():
    
    tricks = []
    def __init__(self, name, age, breed):
        self.name = name
        self.age = age
        self.breed = breed
    
    def woof(self):
        print("Woof! my name is {} and I am {} years old.".format(self.name, self.age))

    def say_breed(self):
        print("My breed is {}".format(self.breed))
        
    def add_trick(self, trick):
        self.tricks.append(trick)
    
```

```python
c1 = Cat('Tom', 5, "Persian")
c1.meow()
c1.say_breed()
```

```python
d1 = Dog('Sammy', 6, 'Husky')
d1.woof()
d1.say_breed()
```

```python
d1.tricks
```

```python
d1.add_trick('Roll Over')
```

```python
d1.tricks
```

### Inheritance

```python
class Person():
    def __init__(self, name, age, gender):
        self.name = name #1st attribute
        self.age = age #2nd attribute
        self.gender = gender #3rd attribute
    
    def introduce(self): #A method
        print('My name is {}, and I am {} years old.'.format(self.name, self.age))
```

```python
class Student(Person):
    
    def introduce(self):
        print('Hahahahaha')
    def who_am_I(self): #Your second method
        print('Hello, my name is {} and I am a student.'.format(self.name))
```

```python
s1 = Student('John', 24, 'Male')
```

```python
s1.age
```

```python
s1.gender
```

```python
s1.name
```

```python
s1.introduce()
```

```python
s1.who_am_I()
```

```python
p1 = Person('John', 24, 'Male')
p1.introduce()
```
