```python
def is_prime(num):
    
    flag = True #Aya prime hast?
    if num == 1:
        flag = False
    elif num == 2:
        flag = True
    
    else:
        for x in range(2, num):
            if (num % x == 0):
                flag = False
                break
                
    print(flag)
```

```python
x = 1349
y = is_prime(x)
```

```python
def my_func10(my_list):
    length = len(my_list)
    M = max(my_list)
    m = min(my_list)
    R = M - m
    S = sum(my_list)
    list_mean = S / length
    
    return S,R, list_mean
```

```python
def my_func10(my_list):
    length = len(my_list)
    M = max(my_list) #The maximum element
    m = min(my_list)
    R = M - m
    S = sum(my_list)
    list_mean = S / length
    my_dict = {'Sum':S , 'Range':R, 'Mean':list_mean,'Max':M, 'Min':m}
    return my_dict
```

```python
my_func10()
```

```python
def my_func10_doc(my_list):
    """
    Calculates some features of an input list.
    
    It calculates the maximum and the minimum and range and sum of all the elements in the list
    
    Parameters:
    my_list (list): Is a list of numbers, ints and floats.
    
    Returns:
    dict: This dictionary contains the sum and the range and the mean and max and min of my_list.
    """
    
    length = len(my_list)
    M = max(my_list) #The maximum element
    m = min(my_list)
    R = M - m
    S = sum(my_list)
    list_mean = S / length
    my_dict = {'Sum':S , 'Range':R, 'Mean':list_mean,'Max':M, 'Min':m}
    return my_dict
```

```python
my_func10_doc()
```
